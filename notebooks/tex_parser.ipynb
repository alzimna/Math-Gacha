```python
%load_ext autoreload
%autoreload 2

import sys
from pathlib import Path
import re
sys.path.append("..")

import pandas as pd

from parser import Problem_to_CSV,Answer_to_CSV

problem_file = Path("../Tex/soal.tex")
problemsv1 = Problem_to_CSV(problem_file).problems_df

answer_file = Path("../Tex/jawaban.tex")
answerv1 = Answer_to_CSV(answer_file).answers_df
```
